## 5. Preparar los nombres de las columnas

Al explorar el dataset notamos dos cosas:

1. Los nombres de las columnas son largos y difíciles de escribir.
2. Los **datos históricos** (`..._variant_estimates`) y las **proyecciones de la ONU** (`..._variant_medium__projected`) vienen en **columnas separadas**. Cuando una fila es histórica, la columna de proyección está vacía (`NaN`), y viceversa.

Vamos a renombrar las columnas y a crear dos columnas unificadas, `births` y `deaths`, que contengan el dato histórico cuando existe y la proyección de OWID cuando no.

````python
df = df.rename(columns={
    "entity": "Entity",
    "code": "Code",
    "year": "Year",
    "births__sex_all__age_all__variant_estimates": "births_estimates",
    "births__sex_all__age_all__variant_medium__projected": "births_projected",
    "deaths__sex_all__age_all__variant_estimates": "deaths_estimates",
    "deaths__sex_all__age_all__variant_medium__projected": "deaths_projected",
})


df.head()
````

````python
df_inicial = df.copy()
df_inicial.head(3)
````

````python
# Columna unificada: dato histórico si existe; si no, la proyección de OWID
df["births"] = df["births_estimates"].fillna(df["births_projected"])
df["deaths"] = df["deaths_estimates"].fillna(df["deaths_projected"])
df.head()
````

# PARTE 5. Separar datos históricos y proyectados

## 10. Identificar el punto de separación

El dataset contiene **datos históricos** y **proyecciones**. Antes de separarlos, verifiquemos dónde termina uno y empieza el otro:

````python
ultimo_historico = df_mexico.loc[df_mexico["births_estimates"].notna(), "Year"].max()
primera_proyeccion = df_mexico.loc[df_mexico["births_projected"].notna(), "Year"].min()

print("Último año con dato histórico:", ultimo_historico)
print("Primer año con proyección de OWID:", primera_proyeccion)
````

# PARTE 6. Preparar las variables para Machine Learning

## 11. Definir X e y

Para nuestro primer modelo utilizaremos:

* **X:** año → la **variable predictora**
* **y:** número de nacimientos → la **variable objetivo**

````python
X = df_historico[["Year"]]
y_nacimientos = df_historico["births"]
````

````python
X.head()
````

````python
y_nacimientos.head()
````

## 12. Convertir los datos a NumPy

Obtenemos los valores como arreglos de NumPy:

````python
X_numpy = X.to_numpy()
y_numpy = y_nacimientos.to_numpy()

print("X_numpy.shape:", X_numpy.shape)
print("y_numpy.shape:", y_numpy.shape)
````

````python
type(X_numpy), type(y_numpy)
````

### Observación

Scikit-learn espera normalmente que las variables predictoras tengan la forma:

```text
(número de observaciones, número de variables)
```
Es decir, que scikit-learn interpreta X como una tabla de datos.

Por eso `X` tiene **dos dimensiones**, mientras que `y` tiene una sola.

En nuestro ejemplo tenemos:

74 observaciones (registros).

1 variable predictora: año (estas son las caracteristicas o atributos usadas para la predicción).

1 variable objetivo: nacimientos (valor que deseamos predecir).

# PARTE 7. Modelo 1 – Regresión lineal

## 13. Crear el modelo

````python
modelo_lineal = LinearRegression()
````

## 14. Entrenar el modelo

Utilizamos los datos históricos para entrenar el modelo.

````python
modelo_lineal.fit(X, y_nacimientos)
````

Una regresión lineal aprende una recta `y = a·x + b`. Veamos qué valores aprendió:

````python
print("Pendiente (a):", modelo_lineal.coef_[0])
print("Intercepto (b):", modelo_lineal.intercept_)
````

## 15. Realizar predicciones

Utilizamos el modelo para estimar los nacimientos correspondientes a los **años históricos**.

````python
predicciones_lineal = modelo_lineal.predict(X)

predicciones_lineal[:10]
````

## 16. Realizar proyecciones hasta 2100

Creamos los años para los que queremos generar **nuestras propias proyecciones**.

````python
anios_futuros = np.arange(2024, 2101).reshape(-1, 1)

print("Forma:", anios_futuros.shape)
anios_futuros[:10]
````

> **Detalle técnico:** entrenamos el modelo con `X`, que es un `DataFrame` con una columna llamada `Year`. Scikit-learn recuerda ese nombre, así que para predecir también debemos entregarle un `DataFrame` con una columna `Year`. Si le pasáramos un arreglo de NumPy "a secas" mostraría una advertencia (*UserWarning: X does not have valid feature names*). Por eso convertimos los años futuros a `DataFrame`.

````python
X_futuro = pd.DataFrame(anios_futuros, columns=["Year"])

X_futuro.head()
````

````python
predicciones_lineal_futuras = modelo_lineal.predict(X_futuro)
````

````python
print(predicciones_lineal_futuras[:10])
````

# PARTE 8. Visualizar la regresión lineal

## 17. Comparar datos históricos y predicciones

````python
plt.figure(figsize=(12, 6))

plt.scatter(
    df_historico["Year"],
    df_historico["births"],
    label="Datos históricos",
    alpha=0.6
)

# Ajuste del modelo sobre los años históricos
plt.plot(
    df_historico["Year"],
    predicciones_lineal,
    color="tab:orange",
    label="Regresión lineal (ajuste histórico)"
)

# Proyección hacia el futuro
plt.plot(
    anios_futuros.flatten(),
    predicciones_lineal_futuras,
    color="tab:orange",
    linestyle="--",
    label="Regresión lineal (proyección)"
)

plt.title("Proyección de nacimientos – Regresión lineal")
plt.xlabel("Año")
plt.ylabel("Nacimientos")

plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````

### Observación

Analiza cómo continúa la tendencia histórica cuando utilizamos una **función lineal**. ¿Qué pasa con los valores hacia el año 2100?

**Respuesta:** La regresión lineal prolonga una recta ascendente: estima un aumento de aproximadamente 8,511 nacimientos por año y llega a unos **3.14 millones de nacimientos en 2100**. Esto ocurre porque resume todo el periodo 1950–2023 con una sola pendiente positiva, aunque los nacimientos hayan disminuido en las décadas recientes. Por eso su proyección no reproduce el descenso de OWID, que llega a aproximadamente 1.10 millones en 2100.


# PARTE 9. Modelo 2 – Regresión polinomial

## 18. Crear variables polinomiales

La regresión lineal solamente puede representar una relación **lineal** entre las variables. Ahora construiremos variables polinomiales para permitir **curvas**. Utilizaremos un polinomio de **grado 2**:

$$y = a \cdot x^2 + b \cdot x + c$$

````python
poly = PolynomialFeatures(
    degree=2,
    include_bias=False
)
````

Transformamos `X`:

````python
X_poly = poly.fit_transform(X)

X_poly[:5]
````

Observa que ahora cada año tiene **dos columnas**: el año (`x`) y el año al cuadrado (`x²`).

## 19. Crear y entrenar el modelo polinomial

Seguimos utilizando `LinearRegression`; lo único que cambió son las **variables de entrada**.

````python
modelo_polinomial = LinearRegression()

modelo_polinomial.fit(
    X_poly,
    y_nacimientos
)
````

## 20. Generar predicciones históricas

````python
predicciones_polinomiales = modelo_polinomial.predict(
    X_poly
)
````

## 21. Generar proyecciones hasta 2100

Transformamos los años futuros utilizando **el mismo objeto `poly`** con el que transformamos los datos de entrenamiento.

````python
anios_futuros_poly = poly.transform(
    X_futuro
)

predicciones_polinomiales_futuras = modelo_polinomial.predict(
    anios_futuros_poly
)
````

# PARTE 10. Visualizar la regresión polinomial

## 22. Graficar las predicciones

````python
plt.figure(figsize=(12, 6))

plt.scatter(
    df_historico["Year"],
    df_historico["births"],
    label="Datos históricos",
    alpha=0.6
)

plt.plot(
    df_historico["Year"],
    predicciones_polinomiales,
    color="tab:green",
    label="Regresión polinomial (ajuste histórico)"
)

plt.plot(
    anios_futuros.flatten(),
    predicciones_polinomiales_futuras,
    color="tab:green",
    linestyle="--",
    label="Regresión polinomial (proyección)"
)

plt.title("Proyección de nacimientos – Regresión polinomial")
plt.xlabel("Año")
plt.ylabel("Nacimientos")

plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````

### Observación

Compara el comportamiento de esta gráfica con el modelo de regresión lineal. ¿Se ajusta mejor a los datos históricos? ¿Qué ocurre con la proyección hacia el año 2100? ¿Los valores que predice tienen sentido para un número de nacimientos?

**Respuesta:** Visualmente, la regresión polinomial de grado 2 representa mejor la curvatura de los datos históricos: primero el crecimiento y después la disminución de los nacimientos. Sin embargo, al prolongar la parábola, el descenso se vuelve cada vez más pronunciado. Las predicciones se hacen negativas desde **2058** y llegan a aproximadamente **−4.40 millones en 2100**. Estos valores no tienen sentido para un número de nacimientos, que no puede ser negativo. Un ajuste histórico más cercano no garantiza una extrapolación razonable.


# PARTE 11. Modelo 3 – Random Forest

## 23. Crear el modelo

Un **Random Forest** está formado por muchos árboles de decisión. Utilizaremos `RandomForestRegressor`.

````python
modelo_rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
````

## 24. Entrenar el modelo

````python
modelo_rf.fit(
    X,
    y_nacimientos
)
````

## 25. Generar predicciones históricas

````python
predicciones_rf = modelo_rf.predict(X)
````

## 26. Generar proyecciones futuras

````python
predicciones_rf_futuras = modelo_rf.predict(
    X_futuro
)
````

# PARTE 12. Visualizar Random Forest

## 27. Graficar las predicciones

````python
plt.figure(figsize=(12, 6))

plt.scatter(
    df_historico["Year"],
    df_historico["births"],
    label="Datos históricos",
    alpha=0.6
)

plt.plot(
    df_historico["Year"],
    predicciones_rf,
    color="tab:red",
    label="Random Forest (ajuste histórico)"
)

plt.plot(
    anios_futuros.flatten(),
    predicciones_rf_futuras,
    color="tab:red",
    linestyle="--",
    label="Random Forest (proyección)"
)

plt.title("Proyección de nacimientos – Random Forest")
plt.xlabel("Año")
plt.ylabel("Nacimientos")

plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````

### Observación

Fíjate en dos cosas: (1) qué tan bien sigue el modelo los datos históricos y (2) qué sucede con la línea después de 2023. ¿Continúa la tendencia? ¿Por qué crees que sucede esto? Investiga.

**Respuesta:** Random Forest sigue de cerca las variaciones históricas, pero después de 2023 produce una línea horizontal: estima aproximadamente **2,054,207 nacimientos por año** durante todo el periodo 2024–2100. Por lo tanto, no continúa la tendencia descendente reciente.

Sus árboles dividen los años mediante umbrales y asignan predicciones constantes a las hojas; el bosque promedia las predicciones de los árboles. La documentación explica este promedio y la dificultad de los árboles para extrapolar. Véanse [RandomForestRegressor](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestRegressor.html) y [árboles de decisión](https://scikit-learn.org/stable/modules/tree.html).

Aplicado a este ejercicio, donde la única variable predictora es el año, todos los años posteriores al máximo de entrenamiento recorren las mismas ramas y terminan en la misma hoja de cada árbol. De ahí se deduce que el promedio del bosque permanece constante, aunque el año siga aumentando.

### 1.
¿Qué diferencia observaste entre la regresión lineal, la regresión polinomial y Random Forest?

**Respuesta:** La regresión lineal resume la relación entre año y variable objetivo mediante una recta, por lo que mantiene una tasa de cambio constante. La regresión polinomial permite una curva y representa cambios de tendencia, pero puede producir valores extremos fuera del periodo histórico. Random Forest sigue más de cerca las variaciones de los datos de entrenamiento, aunque en este ejercicio sus predicciones futuras son constantes. Para nacimientos, la recta sube, la parábola baja hasta valores negativos y el bosque queda horizontal; para defunciones, la recta y la parábola suben a ritmos diferentes y el bosque también queda horizontal.

### 2.
¿Por qué los tres modelos generan predicciones diferentes si utilizaron los mismos datos históricos?

**Respuesta:** Porque cada modelo representa la relación entre las variables de una manera distinta. La regresión lineal ajusta una recta; la polinomial utiliza el año y su cuadrado para ajustar una parábola; Random Forest aprende divisiones de los datos mediante árboles. Aunque reciben los mismos registros históricos, estas estructuras producen funciones diferentes, especialmente cuando se predicen años que no estuvieron en el entrenamiento.

### 3.
¿Qué modelo produjo una tendencia más cercana visualmente a las proyecciones de OWID?

**Respuesta:** Depende de la variable y del horizonte. En nacimientos, la regresión polinomial coincide con la dirección descendente de OWID al inicio, pero después cae demasiado y predice valores negativos. Random Forest está cerca del nivel inicial y queda menos alejado que los otros dos hacia 2100, aunque su línea horizontal no reproduce la tendencia descendente. En defunciones, la polinomial se parece más al crecimiento de OWID en buena parte del tramo inicial e intermedio, pero luego lo sobreestima mucho: llega a unos 3.51 millones frente a 1.79 millones de OWID en 2100. En ese año la lineal, con unos 0.98 millones, queda más cerca. Por ello, no hay un modelo que reproduzca de manera convincente ambas curvas durante todo el periodo.

### 4.
¿El modelo que visualmente parece acercarse más a las proyecciones de OWID necesariamente es el mejor modelo? Explica.

**Respuesta:** No. La semejanza visual depende de la escala de la gráfica y del periodo que se observe. Además, OWID ofrece proyecciones con supuestos e incertidumbre, no los valores reales que necesariamente ocurrirán. Para elegir un modelo habría que evaluar su desempeño con años históricos que no se hayan usado para entrenar, respetando el orden temporal, y utilizar métricas como MAE o RMSE. También hay que revisar que sus predicciones sean demográficamente plausibles. En esta práctica la comparación es visual y todavía no permite declarar un mejor modelo mediante una evaluación formal.

Con el año como única entrada, todos los valores posteriores a 2023 llegan a las mismas hojas de cada árbol, por lo que el promedio del bosque no cambia. Esto explica por qué las líneas quedan horizontales y no prolongan las tendencias recientes.

### 6.
¿Por qué debemos tener cuidado al utilizar un modelo entrenado con datos históricos para realizar predicciones hasta el año 2100?

**Respuesta:** Porque estamos extrapolando hasta 77 años después del último año de entrenamiento. Las relaciones históricas pueden cambiar y el año, por sí solo, no explica factores como la fecundidad, la estructura por edades, la migración o las condiciones de salud. Un buen ajuste al pasado puede producir resultados poco realistas a largo plazo, como los nacimientos negativos del modelo polinomial o las cifras constantes de Random Forest. Debemos considerar la incertidumbre, comprobar el desempeño en periodos posteriores al entrenamiento y utilizar variables y supuestos demográficos adecuados antes de interpretar estas estimaciones como pronósticos confiables.

# Ejercicios nuevos: China y México desde 2000

En los dos ejercicios se usan los tres modelos de la práctica: regresión lineal (RL), regresión polinomial (RP) y Random Forest (RF). Los modelos se entrenan con datos históricos hasta 2023. Las líneas después de 2023 son predicciones de cada modelo.

## 1. Extraer China

Seleccionamos solo las filas del país China.

````python
df_china = df[df["Entity"] == "China"].copy()
df_china = df_china.sort_values("Year")

print("Años de China:", df_china["Year"].min(), "a", df_china["Year"].max())
display(df_china[["Year", "births", "deaths"]].head())
````

### Preparar los datos históricos de China

Solo usamos los años con datos históricos de nacimientos y defunciones. Creamos también los años que aparecerán en las gráficas.

````python
china_historico = df_china[
    (df_china["Year"] <= 2023) &
    df_china["births_estimates"].notna() &
    df_china["deaths_estimates"].notna()
].copy()

X_china = china_historico[["Year"]]
anios_china = pd.DataFrame({"Year": range(int(china_historico["Year"].min()), 2101)})

polinomio_china = PolynomialFeatures(degree=2, include_bias=False)
X_china_polinomial = polinomio_china.fit_transform(X_china)
anios_china_polinomial = polinomio_china.transform(anios_china)

print("Datos históricos de China:", len(china_historico), "años")
````

### China: nacimientos

Entrenamos RL, RP y RF con el número histórico de nacimientos y graficamos sus resultados.

````python
y_china_nacimientos = china_historico["births_estimates"]

china_rl_nacimientos = LinearRegression()
china_rl_nacimientos.fit(X_china, y_china_nacimientos)

china_rp_nacimientos = LinearRegression()
china_rp_nacimientos.fit(X_china_polinomial, y_china_nacimientos)

china_rf_nacimientos = RandomForestRegressor(n_estimators=100, random_state=42)
china_rf_nacimientos.fit(X_china, y_china_nacimientos)

plt.figure(figsize=(12, 6))
plt.scatter(china_historico["Year"], y_china_nacimientos, color="black", s=18, label="Datos históricos")
plt.plot(anios_china["Year"], china_rl_nacimientos.predict(anios_china), label="RL")
plt.plot(anios_china["Year"], china_rp_nacimientos.predict(anios_china_polinomial), label="RP")
plt.plot(anios_china["Year"], china_rf_nacimientos.predict(anios_china), label="RF")
plt.title("China: modelos para nacimientos")
plt.xlabel("Año")
plt.ylabel("Nacimientos")
plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````

### China: defunciones

Ahora entrenamos los mismos modelos con el número histórico de defunciones.

````python
y_china_defunciones = china_historico["deaths_estimates"]

china_rl_defunciones = LinearRegression()
china_rl_defunciones.fit(X_china, y_china_defunciones)

china_rp_defunciones = LinearRegression()
china_rp_defunciones.fit(X_china_polinomial, y_china_defunciones)

china_rf_defunciones = RandomForestRegressor(n_estimators=100, random_state=42)
china_rf_defunciones.fit(X_china, y_china_defunciones)

plt.figure(figsize=(12, 6))
plt.scatter(china_historico["Year"], y_china_defunciones, color="black", s=18, label="Datos históricos")
plt.plot(anios_china["Year"], china_rl_defunciones.predict(anios_china), label="RL")
plt.plot(anios_china["Year"], china_rp_defunciones.predict(anios_china_polinomial), label="RP")
plt.plot(anios_china["Year"], china_rf_defunciones.predict(anios_china), label="RF")
plt.title("China: modelos para defunciones")
plt.xlabel("Año")
plt.ylabel("Defunciones")
plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````

```python
import pandas as pd
import requests

# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/life-expectancy.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/life-expectancy.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
df.head()
```

```python
#Renombrar columnas
df.rename(columns={"entity": "Pais"}, inplace=True)
df.rename(columns={"code": "codigo"}, inplace=True)
df.rename(columns={"year": "años"}, inplace=True)
df.rename(columns={"life_expectancy_0": "Esperanza de vida"}, inplace=True)
```

```python
df.head()
```

```python
df.set_index("Pais")
```

```python
print(df.columns.tolist())
```

```python
df.columns = [
    "Pais",
    "codigo",
    "años",
    "Esperanza de vida"
]
```

```python
df.loc[df["Pais"].str.contains("U", case = False, regex = True)]
```

```python
df.dropna(subset=["codigo"], inplace=True)
df.isnull().sum()
```

```python
resultado = (
    df.groupby("Pais")["Esperanza de vida"]
      .mean()
      .sort_values(ascending=True)
)
print(resultado)
```

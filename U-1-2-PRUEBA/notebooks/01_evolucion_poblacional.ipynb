## Actividad 3 — Estadística descriptiva

**Propósito:** aplicar medidas estadísticas básicas a una variable poblacional.

Se selecciona el año histórico **2020** y se analiza la variable `Annual change in population`. Se eliminan únicamente los valores faltantes de esta variable antes de calcular los estadísticos.

```python
variable = "Annual change in population"
df_2020 = df.loc[(df["Year"] == 2020) & df[variable].notna()].copy()
print("Registros históricos de 2020:", len(df_2020))
df_2020[["Entity", "Year", variable]].head()
```

```python
cambio_2020 = df_2020[variable]

estadisticos_2020 = pd.Series({
    "Media": cambio_2020.mean(),
    "Mediana": cambio_2020.median(),
    "Varianza": cambio_2020.var(),
    "Desviación estándar": cambio_2020.std(),
    "Mínimo": cambio_2020.min(),
    "Máximo": cambio_2020.max(),
}, name="Cambio anual de población en 2020")

estadisticos_2020.to_frame().style.format("{:,.2f}")
```

```python
fila_minimo = df_2020.loc[cambio_2020.idxmin(), ["Entity", variable]]
fila_maximo = df_2020.loc[cambio_2020.idxmax(), ["Entity", variable]]

entidades_extremas = pd.DataFrame({
    "Mínimo": fila_minimo,
    "Máximo": fila_maximo,
}).T
entidades_extremas
```

```python
# Comprobación de la media y la mediana con NumPy
print("Media con NumPy:", np.mean(cambio_2020))
print("Mediana con NumPy:", np.median(cambio_2020))
```

### Respuestas de la Actividad 3

1. **¿Cuál es el valor de la media?** La media es **2,124,908.85 habitantes**.
2. **¿Cuál es el valor de la mediana?** La mediana es **34,293.50 habitantes**.
3. **¿Cuál es el valor de la desviación estándar?** La desviación estándar muestral es **9,522,989.82 habitantes**.
4. **¿Qué entidad presenta el valor mínimo y cuál el máximo?** **Venezuela** presenta el mínimo, con **-494,016 habitantes**, y **World** presenta el máximo, con **75,707,590 habitantes**. `World` aparece porque el dataset incluye agregados además de países.

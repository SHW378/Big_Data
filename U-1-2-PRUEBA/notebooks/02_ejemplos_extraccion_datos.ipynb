# Ejemplos de extracción de datos

Ejemplos independientes para practicar la selección de filas y columnas con pandas usando el conjunto de evolución poblacional.

```python
from pathlib import Path
import pandas as pd
```

```python
# Funciona si el notebook se abre desde la raíz del proyecto o desde la carpeta notebooks.
raiz = Path.cwd() if (Path.cwd() / "data").exists() else Path.cwd().parent
ruta_datos = raiz / "data" / "raw" / "annual-population-growth" / "annual-population-growth.csv"
df = pd.read_csv(ruta_datos)
df.head()
```

```python
# 1. Extraer las columnas principales
columnas_principales = df[["Entity", "Code", "Year", "Annual change in population"]]
columnas_principales.head(5)
```

```python
# 2. Extraer todos los registros de México
datos_mexico = df.loc[df["Entity"] == "Mexico", ["Year", "Annual change in population"]]
datos_mexico.head(10)
```

```python
# 3. Extraer datos de México entre 2000 y 2020
mexico_2000_2020 = df.loc[
    (df["Entity"] == "Mexico") & df["Year"].between(2000, 2020),
    ["Entity", "Year", "Annual change in population"]
]
mexico_2000_2020.head(10)
```

```python
# 4. Comparar varios países en un año específico
paises = ["Mexico", "United States", "Brazil", "Canada"]
comparacion_2020 = df.loc[
    df["Entity"].isin(paises) & (df["Year"] == 2020),
    ["Entity", "Code", "Year", "Annual change in population"]
]
comparacion_2020
```

```python
# 5. Extraer países con disminución de población en 2020
es_pais = df["Code"].str.len().eq(3)
disminucion_2020 = df.loc[
    es_pais & (df["Year"] == 2020) & (df["Annual change in population"] < 0),
    ["Entity", "Annual change in population"]
].sort_values("Annual change in population")
disminucion_2020.head(10)
```

```python
# 6. Extraer los 10 mayores aumentos de población en 2020
mayores_aumentos_2020 = (
    df.loc[es_pais & (df["Year"] == 2020), ["Entity", "Annual change in population"]]
      .nlargest(10, "Annual change in population")
)
mayores_aumentos_2020
```

```python
# 7. Extraer proyecciones disponibles a partir de 2025
proyecciones = df.loc[
    (df["Year"] >= 2025) & df["Annual population change (Projected)"].notna(),
    ["Entity", "Year", "Annual population change (Projected)"]
]
proyecciones.head(10)
```

```python
# 8. Extraer una muestra aleatoria reproducible
muestra = df[["Entity", "Year", "Annual change in population"]].sample(5, random_state=42)
muestra
```
